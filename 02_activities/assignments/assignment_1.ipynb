### Part 1: Building the base Anagram Checker

Given two valid strings, check to see if they are anagrams of each other. If it is, return `True`, else `False`. For this part, we can assume that uppercase letters are the same as if it was a lowercase character.

Examples of anagrams:
* Silent and Listen
* Night and Thing

Example outputs:
```python
anagram_checker("Silent", "listen") # True
anagram_checker("Silent", "Night") # False
anagram_checker("night", "Thing") # True
```

````python
# For testing purposes, we will write our code in the function
def anagram_checker(word_a, word_b):
  # make letters lower case in word_a and word_b for easier comparison
  word_a = word_a.lower()
  word_b = word_b.lower()
  # Check if word_a and word_b are the same length
  if len(word_a) != len(word_b):
    # If the words are of different lengths, return the following line
    return "Word lengths do not match :( "
  # If the words are the same length, sort the words alphabetically and see if they match to one another
  if sorted(word_a) == sorted(word_b):
    # If they match, return the following line
    return "True"
  else:
    # If they do not match, return the following line
    return "False"


# Run your code to check using the words below:
anagram_checker("Silent", "listen")
````
